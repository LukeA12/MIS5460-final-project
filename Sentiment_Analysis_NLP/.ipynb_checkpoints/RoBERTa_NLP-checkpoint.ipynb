```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm.notebook import tqdm

# Set Styles for plots
plt.style.use('ggplot')
```

```python
# Read in data from csv
df = pd.read_csv('Reviews.csv')
df.head()
```

```python
# Scale down the dataset to first 500
print(df.shape)
df = df.head(500)
print(df.shape)
```

```python
# Get a count of the scores and plot on bar chart
ax = df['Score'].value_counts().sort_index() \
    .plot(kind='bar', 
          title='Review Score Counts', 
          figsize=(10,5))

ax.set_xlabel('Review Score')
ax.set_ylabel('Number of Reviews')
plt.show()
```

```python
from transformers import AutoTokenizer
from transformers import AutoModelForSequenceClassification
from scipy.special import softmax
```

```python
MODEL = f"cardiffnlp/twitter-roberta-base-sentiment"
tokenizer = AutoTokenizer.from_pretrained(MODEL)
model = AutoModelForSequenceClassification.from_pretrained(MODEL)
```

```python
# Get one review for an example
example = df['Text'][50]
print(example)
```

```python
# Run on RoBERTa model
encoded_text = tokenizer(example, return_tensors='pt')
output = model(**encoded_text)
scores = output[0][0].detach().numpy()

# Apply softmax to get probabilities
scores = softmax(scores)
print(scores)
```

```python
# Function to get polarity scores from text
def polarity_scores_roberta(example):
    encoded_text = tokenizer(example, return_tensors='pt')
    output = model(**encoded_text)
    scores = output[0][0].detach().numpy()
    scores = softmax(scores)
    scores_dict = {
            'negative': scores[0],
            'neutral': scores[1],
            'positive': scores[2]
            }
    
    return scores_dict
```

```python
# Go over entire dataset
result = {}
for i, row in tqdm(df.iterrows(), total=len(df)):
    try:
        text = row['Text']
        id = row['Id']
        scores = polarity_scores_roberta(text)

        # Store in 
        result[id] = scores

    except Exception as e:
        print(f"Skipped id: {id}, Error: {e}")
```

```python
# Convert results to dataframe
result_df = pd.DataFrame(result).T
result_df = result_df.reset_index().rename(columns={'index':'Id'})
result_df = result_df.merge(df, how='left', on='Id')

# View results
result_df.head()
```

```python
# Plot each Positive, Neutral, and Negative Score for each Star Rating
fig, axes = plt.subplots(1, 3, figsize=(15, 5))
sns.barplot(data=result_df, x='Score', y='positive', ax=axes[0])
sns.barplot(data=result_df, x='Score', y='neutral', ax=axes[1])
sns.barplot(data=result_df, x='Score', y='negative', ax=axes[2])
axes[0].set_title('Positive')
axes[1].set_title('Neutral')
axes[2].set_title('Negative')
plt.show()
```
