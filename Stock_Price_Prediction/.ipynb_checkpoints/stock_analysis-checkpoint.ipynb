```python
import kagglehub

# Download latest version of the dataset
path = kagglehub.dataset_download("aaron7sun/stocknews")

print("Path to dataset files:", path)
```

```python
import pandas as pd
import os
import matplotlib.pyplot as plt

file_path1 = os.path.join(path, "Combined_News_DJIA.csv")
file_path2 = os.path.join(path, "upload_DJIA_table.csv")

# I will start with simple ARIMA modeling using just closing price
# Load both CSVs
news_df = pd.read_csv(file_path1)
stock_df = pd.read_csv(file_path2)

# Convert Date to datetime format
stock_df['Date'] = pd.to_datetime(stock_df['Date'])
news_df['Date'] = pd.to_datetime(news_df['Date'])

# Sort by date (important for time series)
stock_df = stock_df.sort_values('Date')

# Inspect
print(stock_df.head())
print(stock_df.columns)
```

```python
plt.figure(figsize=(10,4))
plt.plot(stock_df['Date'], stock_df['Close'])
plt.title("DJIA Closing Price Over Time")
plt.xlabel("Date")
plt.ylabel("Close Price")
plt.show()

```

```python
# First try
from statsmodels.tsa.arima.model import ARIMA
import matplotlib.pyplot as plt
import pandas as pd

# Date to datetime and set as index
if 'Date' in stock_df.columns:
    stock_df['Date'] = pd.to_datetime(stock_df['Date'])
    stock_df.set_index('Date', inplace=True)
else:
    stock_df.index = pd.to_datetime(stock_df.index)

# Use Close as time series
series = stock_df['Close']

# Split into train/test (e.g. 80/20)
train_size = int(len(series) * 0.8)
train, test = series[:train_size], series[train_size:]

# Fit ARIMA
# AR (p=5): Looks back 5 days to predict the next
# I (d=1): Differentiates once to remove trend
# MA (q=1): Uses past forecast errors to improve predictions.
model = ARIMA(train, order=(5,1,1))
model_fit = model.fit()

# Forecast
forecast = model_fit.forecast(steps=len(test))

# Plot with proper date axis
plt.figure(figsize=(12,6))
plt.plot(train.index, train, label='Training Data', color='blue')
plt.plot(test.index, test, label='Actual Test Data', color='black')
plt.plot(test.index, forecast, label='Forecasted Values', color='red', linestyle='--')

plt.title('ARIMA Forecast of DJIA Close Prices')
plt.xlabel('Date (Years)')
plt.ylabel('DJIA Close Price')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Second try: Instead of forecasting the entire future at once, this refits the ARIMA model at each time step 
# using all available data up to that point.
# This is slower
from statsmodels.tsa.arima.model import ARIMA
import matplotlib.pyplot as plt
import pandas as pd

if 'Date' in stock_df.columns:
    stock_df['Date'] = pd.to_datetime(stock_df['Date'])
    stock_df.set_index('Date', inplace=True)
else:
    stock_df.index = pd.to_datetime(stock_df.index)

# Use 'Close' as the target time series
series = stock_df['Close']

# Split into train/test sets (e.g., 80/20)
train_size = int(len(series) * 0.8)
train, test = series[:train_size], series[train_size:]

# Rolling (Walk-Forward) Forecast Setup
# Initialize history with training data and list to store predictions
history = list(train)
predictions = []

# Walk-forward validation loop
# Each step: fit ARIMA on current history, forecast one step, append result, then add true test value to history
for t in range(len(test)):
    model = ARIMA(history, order=(5,1,1))
    model_fit = model.fit()
    yhat = model_fit.forecast()[0]     # One-step forecast
    predictions.append(yhat)
    actual = test.iloc[t]
    history.append(actual)             # Add actual test value for next round

# --- Plot Results ---
plt.figure(figsize=(12,6))
plt.plot(train.index, train, label='Training Data', color='blue')
plt.plot(test.index, test, label='Actual Test Data', color='black')
plt.plot(test.index, predictions, label='Rolling Forecast', color='red', linestyle='--')

plt.title('Rolling ARIMA Forecast vs Actual DJIA Close Prices')
plt.xlabel('Date (Years)')
plt.ylabel('DJIA Close Price')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
from prophet import Prophet
import pandas as pd
import matplotlib.pyplot as plt
from textblob import TextBlob

news_df = pd.read_csv(file_path1)
stock_df = pd.read_csv(file_path2)

# Preprocessing the news dataset (Combined_News_DJIA.csv) so I can compute a single daily sentiment score for each trading day.
headline_cols = [col for col in news_df.columns if 'Top' in col]
news_df['all_headlines'] = news_df[headline_cols].astype(str).agg(' '.join, axis=1)

# This computes a sentiment polarity value between -1 and +1 for that daily combined text.
# -1 = very negative news
# 0 = neutral
# +1 = very positive news
news_df['sentiment'] = news_df['all_headlines'].apply(lambda x: TextBlob(x).sentiment.polarity)

# Keep only Date and sentiment
sent_df = news_df[['Date', 'sentiment']].copy()
sent_df['Date'] = pd.to_datetime(sent_df['Date'])

# If multiple news per day, average them
sent_df = sent_df.groupby('Date', as_index=False)['sentiment'].mean()

# Prepare stock data
stock_df['Date'] = pd.to_datetime(stock_df['Date'])
stock_df = stock_df[['Date', 'Close']]

# Merge by date
merged_df = pd.merge(stock_df, sent_df, on='Date', how='inner')

# Prophet requires columns ds (date) and y (target)
prophet_df = merged_df.rename(columns={'Date': 'ds', 'Close': 'y'})

# Train Prophet with sentiment as regressor
model = Prophet()
model.add_regressor('sentiment')
model.fit(prophet_df)

# Create future dates
future = model.make_future_dataframe(periods=30)

# Merge with sentiment again
future = pd.merge(future, sent_df, left_on='ds', right_on='Date', how='left')
future.drop(columns='Date', inplace=True)

# Fill missing future sentiment values (e.g., use last known sentiment)
future['sentiment'] = future['sentiment'].ffill().bfill()

# Predict
forecast = model.predict(future)

# Plot
fig1 = model.plot(forecast)
plt.title('Facebook Prophet Stock Price Forecast (with News Sentiment)')
plt.xlabel('Date')
plt.ylabel('DJIA Close Price')
# Adding legend manually
plt.legend(
    ['Observed (Actual Data)', 'Predicted (Prophet Forecast)', 'Uncertainty Interval'],
    loc='upper left'
)
plt.show()

# There are four component graphs
# 1. Trend: shows the long-term movement Prophet found in the stock prices
# 2. Weekly seasonality: shows the trend of the day of week vs relative impact on the price
# 3. Yearly seasonality: shows the trend of the time of the year vs relative impact on the price
# 4. Impact of news sentiment: shows how much the additive effect of sentiment contributed to the forecast on each day from 2009 to 2016
# Y-axis (-7.5 to 10): how much the sentiment regressor adds/subtracts from the baseline price forecast
# X-axis: when those affects occurred in time

fig2 = model.plot_components(forecast)
plt.show()
```
